## 1. Representación del problema (repartidor)
 
El problema consiste en llevar un pedido desde el Almacén hasta la Zona Norte, pudiendo pasar por puntos intermedios de la ciudad.
 
| Elemento del Problema | Descripción |
| --- | --- |
| **Estado inicial** | el repartidor se encuentra en el Almacén. |
| **Estado objetivo** | el repartidor llega a la Zona Norte. |
| **Acciones posibles** | moverse de un punto a otro punto que esté directamente conectado con él. |
| **Costo de cada acción** | 1 paso (costo unitario por cada tramo o trayecto recorrido) |
---
 

## 2. Representación del mapa como grafo
 
Se usa un diccionario de Python, donde cada clave es un punto de la ciudad y cada valor es la lista de puntos conectados directamente con él.
 

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

## 3. Implementación de BFS
 
BFS explora el grafo por niveles, usando una cola (FIFO). Por eso primero revisa todos los vecinos directos del Almacén antes de avanzar más lejos.
 

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]          # cola de caminos, empieza con el estado inicial
    visitados = set()
 
    while cola:
        camino = cola.pop(0)   # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]
 
        if nodo_actual == objetivo:
            return camino
 
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)
 
    return None  # no se encontro un camino
 
 
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```
