## 1. Representación del problema (repartidor)
 
El problema consiste en llevar un pedido desde el Almacén hasta la Zona Norte, pudiendo pasar por puntos intermedios de la ciudad.
 
| Elemento del Problema | Descripción |
| --- | --- |
| **Estado inicial** | el repartidor se encuentra en el Almacén. |
| **Estado objetivo** | el repartidor llega a la Zona Norte. |
| **Acciones posibles** | moverse de un punto a otro punto que esté directamente conectado con él. |
| **Costo de cada acción** | 1 paso (costo unitario por cada tramo o trayecto recorrido) |
---
 

## 2. Representación del mapa como grafo
 
Se usa un diccionario de Python, donde cada clave es un punto de la ciudad y cada valor es la lista de puntos conectados directamente con él.
