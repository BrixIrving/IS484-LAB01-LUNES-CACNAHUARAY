# Representación de Problemas, Búsqueda y Fundamentos Matemáticos de IA
  
Este notebook desarrolla las actividades de la Práctica de Laboratorio 002: representación de un problema como grafo, búsqueda en amplitud (BFS) y en profundidad (DFS), y el uso de NumPy para trabajar con vectores, matrices y distancias entre observaciones.
 
---
 

## 1. Representación del problema (repartidor)
 
El problema consiste en llevar un pedido desde el Almacén hasta la Zona Norte, pudiendo pasar por puntos intermedios de la ciudad.
 
| Elemento del Problema | Descripción |
| --- | --- |
| **Estado inicial** | el repartidor se encuentra en el Almacén. |
| **Estado objetivo** | el repartidor llega a la Zona Norte. |
| **Acciones posibles** | moverse de un punto a otro punto que esté directamente conectado con él. |
| **Costo de cada acción** | 1 paso (costo unitario por cada tramo o trayecto recorrido) |
---
 

## 2. Representación del mapa como grafo
 
Se usa un diccionario de Python, donde cada clave es un punto de la ciudad y cada valor es la lista de puntos conectados directamente con él.
 

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

## 3. Implementación de BFS
 
BFS explora el grafo por niveles, usando una cola (FIFO). Por eso primero revisa todos los vecinos directos del Almacén antes de avanzar más lejos.
 

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]          # cola de caminos, empieza con el estado inicial
    visitados = set()
 
    while cola:
        camino = cola.pop(0)   # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]
 
        if nodo_actual == objetivo:
            return camino
 
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)
 
    return None  # no se encontro un camino
 
 
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

 
BFS encuentra el camino **Almacén → Centro → Zona Norte**, que tiene 2 pasos. Esto tiene sentido porque BFS revisa primero todos los vecinos cercanos antes de ir más lejos, así que apenas encuentra la Zona Norte en el segundo nivel, se detiene ahí sin seguir explorando caminos más largos.
 
---
 

## 4. Implementación de DFS
 
DFS explora un camino hasta el final antes de retroceder, usando una pila (LIFO).
 

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """
    pila = [[inicio]]
    visitados = set()
 
    while pila:
        camino = pila.pop()    # sacar el ULTIMO camino de la pila (LIFO)
        nodo_actual = camino[-1]
 
        if nodo_actual == objetivo:
            return camino
 
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)
 
    return None
 
 
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

 
### Comparación BFS vs. DFS
 
| Algoritmo | Camino encontrado | Pasos |
|---|---|---|
| BFS | Almacén → Centro → Zona Norte | 2 |
| DFS | Almacén → Terminal → Mercado → Zona Norte | 3 |
 
No es el mismo camino. BFS encuentra el camino más corto (2 pasos) porque revisa nivel por nivel y se detiene en cuanto encuentra la meta en el nivel más cercano posible. DFS, en cambio, se mete por la primera rama que encuentra (Terminal, porque fue el último vecino agregado a la pila) y sigue por ahí hasta toparse con la Zona Norte, sin fijarse si existía un camino más corto por otro lado. Por eso DFS terminó dando una "vuelta más larga", tal como se explica en el material de apoyo del algoritmo.

## 5. Vectores y matrices con NumPy
 
Se representan cuatro observaciones de temperatura y humedad, cada una como un vector, y se juntan en una matriz (dataset).
 

```python
import numpy as np
 
obs1 = np.array([22.5, 60])   # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
 
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

 
Cada fila es una observación (un momento distinto de medición) y cada columna es una característica (temperatura o humedad). Esto es justo la estructura típica de un dataset.
 

```python
print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```
